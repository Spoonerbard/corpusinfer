```python
import os
from pathlib import Path 
import json
# NOTE: use pandas for greater performance?
from collections import Counter
import re

# root directory of the project
PROJECT_ROOT = Path.cwd().parent
output_dir = PROJECT_ROOT / "data" / "output"
output_dir.mkdir(parents=True, exist_ok=True)
# file path
shakespeare_text_path = PROJECT_ROOT / "data" / "complete_shakespeare.txt"
```

```python
def count_words_in_file(file_path):
    # TODO - read in file, line by line
    with Path(file_path).open("r", encoding="utf-8") as file:
        lines = file.readlines()

    # TODO - setup dictionary to store counts
    word_counts = {}

    # TODO - iterate over each line in the file,
    for line in lines:
        # TODO - iterate over each word in the line
        words = line.lower().split()

        # TODO - repeat for all lines in the file
        for word in words:
            # TODO - check if the word is already in the dictionary
            if word in word_counts:
                # TODO - if it is, increment the count
                word_counts[word] += 1
            else:
                # TODO - if it isn't, add it to the dictionary with a count of 1
                word_counts[word] = 1

    # TODO - update the dictionary with the word counts
    # (The dictionary is updated as words are counted above.)
    output_path = output_dir / "shakespeare_word_counts.json"
    with output_path.open("w", encoding="utf-8") as f:
        json.dump(word_counts, f, indent=4)
    # TODO - print the final word counts
    print('Word counts have been written to:', output_path)
    return word_counts
```

```python

# Function to count words in a file using a Counter and regular expressions
WORD_RE = re.compile(r"[a-z']+")

def count_words_counter(file_path: Path):
    counts = Counter()

    with Path(file_path).open("r", encoding="utf-8") as file:
        for line in file:
            counts.update(WORD_RE.findall(line.lower()))

    return dict(sorted(counts.items(),
                           key=lambda kv: kv[1],
                           reverse=True))
counts = count_words_counter(shakespeare_text_path)
output_path = output_dir / "shakespeare_word_counts_counter.json"
with output_path.open("w", encoding="utf-8") as f:
    json.dump(dict(counts), f, indent=4)
print("Word counts have been written to:", output_path)
```

```python
# file path
shakespeare_text_path = PROJECT_ROOT / "data" / "shakespeare.txt"
# Count words in the Shakespeare text file using the Counter-based function
count_words_counter(shakespeare_text_path)
```
